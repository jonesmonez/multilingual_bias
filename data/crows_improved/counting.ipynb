```python
import os
import contextlib
import io
import pandas as pd

# from bias_bench.test.automated_test import test_crows

directory = "."

to_test_types = ["gender", "race-color", "religion"]
# to_test_languages = ["de_DE", "mt_MT"]
to_test_languages = ["ar_DZ", "ca_ES", "de_DE", "en_US", "es_AR", "fr_FR", "it_IT", "mt_MT", "zh_CN"]

res_dict = {}

for lang in to_test_languages:
    lang_dict = {}
    df = pd.read_csv(f"crows_{lang}.csv")
    types = df["bias_type"].unique()
    for t in types:
        lang_dict[t] = int((df["bias_type"] == t).sum())
        
    
    res_dict[lang] = lang_dict
    
df = pd.DataFrame(res_dict)
df.loc["Summe"] = df.sum()
df["Summe"] = df.sum(axis=1)

    
display(df)
```
