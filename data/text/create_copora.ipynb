```python
from create_corpora import CorporaCreator
creator = CorporaCreator()
```

```python
languages = ["ar_DZ", "ca_ES", "de_DE", "es_ES", "fr_FR", "it_IT", "zh_CN"]

for lang in languages:
    sample10, sample25 = creator.create_samples(lang)

    with open(
        f"{lang}_2.5pct.txt",
        "w",
        encoding="utf-8"
    ) as f:
        f.write("\n\n".join(sample25))


    with open(
        f"{lang}_10pct.txt",
        "w",
        encoding="utf-8"
    ) as f:
        f.write("\n\n".join(sample10))
```
