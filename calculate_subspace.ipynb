```python
from experiments.modules.calculate_debias_subspace import SentenceDebiasWrapper, DensrayDebiasWrapper
```

```python
sent_runner = SentenceDebiasWrapper(save_result=True)
```

```python
languages = ["es_ES", "fr_FR", "it_IT", "mt_MT", "zh_CN"]

for lang in languages:
    print(lang)
    if lang == "mt_MT":
        path_dataset = "data/text/mt_MT_100pct.txt"
    else:
        path_dataset = f"data/text/{lang}_2.5pct.txt"

    if lang == "es_ES":
        attribute_lang = "es_AR"
    else:
        attribute_lang = lang
    sent_runner.setup_data(
        path_to_bias_attributes=f"data/bias_attribute/translated/bias_attribute_words_{attribute_lang}.json",
        lang_debias=lang,
        path_to_dataset=path_dataset,
        n_max_sent=None
    )
    sent_runner.compute_gender_subspace(save_path=".")
    sent_runner.compute_racecolor_subspace(save_path=".")
    sent_runner.compute_religion_subspace(save_path=".")
```
